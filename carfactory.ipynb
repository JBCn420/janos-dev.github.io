```python
import duckdb

# Megszámoljuk, hol hány NULL érték van
query_audit = """
SELECT
    COUNT(*) as osszes_sor,
    COUNT(part_name) as kitoltott_nevek,
    COUNT(*) - COUNT(part_name) as hianyzo_nevek,
    COUNT(price_eur) as kitoltott_arak,
    COUNT(*) - COUNT(price_eur) as hianyzo_arak,
    COUNT(*) - COUNT(supplier) as hianyzo_beszallitok
FROM df_parts
"""
duckdb.query(query_audit).to_df()
```

# Hibás sorok kiiratása NULL/DISTINCT/ROW_NUMBER

```python
import duckdb

#------------Megnézzük szerepelnek e NULL értékek majd kiszürjük őket--------------

# A legegyszerűbb lekérdezés: "Válassz ki mindent az alkatrész táblából"
query_all = """
SELECT *,
    CASE
        WHEN part_name IS NULL OR price_eur IS NULL OR supplier IS NULL
        THEN '⚠️ HIÁNYOS ADAT'
        ELSE '✅ RENDBEN'
    END AS statusz
FROM df_parts
ORDER BY statusz ASC -- Így a hibásak kerülnek előre
"""

# Futtatás és az eredmény megjelenítése
df_eredmeny = duckdb.query(query_all).to_df()
df_eredmeny
```

```python
import duckdb

# Ellenőrizzük mindhárom táblát az egyedi kulcsaik alapján
query_triple_check = """
SELECT 'Alkatrészek' as tabla, COUNT(part_id) - COUNT(DISTINCT part_id) as duplikalt_id_szam FROM df_parts
UNION ALL
SELECT 'Gyártók', COUNT(part_id) - COUNT(DISTINCT part_id) FROM df_manufacturers
UNION ALL
SELECT 'Márkák', COUNT(brand_name) - COUNT(DISTINCT brand_name) FROM df_brands
"""

duckdb.query(query_triple_check).to_df()
```

```python
import duckdb

# Ellenőrizzük egy tablaban a DISTINCet - ez azt jelenti hogy ezekben NINCSEN duplikacio-------------------

query_tabla_check = """

SELECT
COUNT(DISTINCT part_id) as NEMduplikalt_part_id_szamok
FROM df_manufacturers

"""

duckdb.query(query_tabla_check).to_df()
```

```python
import duckdb

# Megkeressük azokat a sorokat, amiknek az ID-ja többször szerepel
query_teljes_duplo = """
SELECT *
FROM df_manufacturers
WHERE part_id IN (
    SELECT part_id
    FROM df_manufacturers
    GROUP BY part_id
    HAVING COUNT(*) > 1
)
ORDER BY part_id
"""

df_duplak = duckdb.query(query_teljes_duplo).to_df()
df_duplak
```

```python
import duckdb

# SQL megoldás: Csoportosítunk ID szerint, és minden csoportból csak egyet tartunk meg
query_clean_sql = """
SELECT part_id, origin_country, warranty_years
FROM df_manufacturers
GROUP BY part_id, origin_country, warranty_years
"""

df_manufacturers_clean = duckdb.query(query_clean_sql).to_df()
print(f"Tisztított sorok száma: {len(df_manufacturers_clean)}")
df_manufacturers_clean.head()
```

```python
import duckdb

# A "profi" tisztító lekérdezés A WHERE sorrend 1 kiirja az egyediket a 2 kiirja a duplikaltakat
query_row_number = """
SELECT
    part_id,
    origin_country,
    warranty_years
FROM (
    SELECT *,
        ROW_NUMBER() OVER(PARTITION BY part_id ORDER BY part_id) as sorrend
    FROM df_manufacturers
)
WHERE sorrend = 1
ORDER BY part_id
"""

# Futtatás és mentés
df_manufacturers_clean = duckdb.query(query_row_number).to_df()

print(f"✅ Tisztítás kész! Sorok száma: {len(df_manufacturers_clean)}")
df_manufacturers_clean.head()
```

# Megtisztitott adatbazis mentése

```python
import duckdb

# SQL-alapú tisztítás: A DISTINCT kulcsszóval kiszűrjük az egyedi sorokat
# és egyből az eredményt tesszük be a df_manufacturers_clean változóba
query_tisztitas = """
SELECT DISTINCT * FROM df_manufacturers
ORDER BY part_id
"""

# A lekérdezés futtatása és Pandas DataFrame-mé alakítása
df_manufacturers_clean = duckdb.query(query_tisztitas).to_df()

# Ellenőrzés SQL-lel: Megszámoljuk az egyedi sorokat
query_count = "SELECT COUNT(*) FROM df_manufacturers_clean"
tiszta_szam = duckdb.query(query_count).fetchone()[0]

print(f"✅ SQL-lel megtisztítva! Maradt összesen: {tiszta_szam} sor.")

# Mentés (ezt Pythonnal csináljuk, mert az a legegyszerűbb fájlkezelés)
df_manufacturers_clean.to_csv('gyartok_tisztitott.csv', index=False)

df_manufacturers_clean.head()
```

```python
import duckdb

# A legegyszerűbb lekérdezés: "Válassz ki mindent az alkatrész táblából"
query_all = """

SELECT
   DISTINCT p.part_name,
    p.price_eur,
    m.origin_country,
    m.warranty_years

FROM df_parts p
LEFT JOIN df_manufacturers_clean m ON p.part_id = m.part_id
LEFT JOIN df_brands b ON m.origin_country = b.origin_country
WHERE p.price_eur > 50
ORDER BY p.price_eur DESC

"""

# Futtatás és az eredmény megjelenítése
df_eredmeny = duckdb.query(query_all).to_df()
df_eredmeny
```

```python
import duckdb

# A legegyszerűbb lekérdezés: "Válassz ki mindent az alkatrész táblából"
query_all = """

SELECT  p.part_name as alkatrész, m.origin_country as Származási_ország
FROM df_parts p
LEFT JOIN df_manufacturers_clean m ON p.part_id = m.part_id
WHERE m.origin_country = 'Svédország'

"""

# Futtatás és az eredmény megjelenítése
df_eredmeny = duckdb.query(query_all).to_df()
df_eredmeny
```

```python
#@title 🔍 Alkatrész kereső ország szerint
valasztott_orszag = "Németország" #@param ["Németország", "Svédország", "USA", "Franciaország", "Japán"]

# Az SQL lekérdezés egyszerűen használja a fenti változót
query = f"""
SELECT p.part_name, p.price_eur, m.origin_country
FROM df_parts p
JOIN df_manufacturers_clean m ON p.part_id = m.part_id
WHERE m.origin_country = '{valasztott_orszag}'
"""

# Futtatás és kiírás
df_eredmeny = duckdb.query(query).to_df()
df_eredmeny
```

```python
import pandas as pd
import duckdb
import ipywidgets as widgets
from IPython.display import display, clear_output

# 1. Kigyűjtjük az egyedi országokat a menühöz
countries = df_manufacturers_clean['origin_country'].unique().tolist()
countries.sort() # Legyenek ABC sorrendben

# 2. Létrehozzuk a legördülő menüt (Dropdown)
dropdown = widgets.Dropdown(
    options=countries,
    description='Ország:',
    disabled=False,
)

# 3. Létrehozunk egy kimeneti ablakot a táblázatnak
output = widgets.Output()

# 4. A függvény, ami lefut, ha váltasz a menüben
def on_change(change):
    if change['type'] == 'change' and change['name'] == 'value':
        with output:
            clear_output(wait=True) # Törli az előző táblázatot
            selected_country = change['new']

            # Az SQL lekérdezés, amibe beillesztjük a választott országot
            query = f"""
            SELECT p.part_name, p.price_eur, m.origin_country
            FROM df_parts p
            JOIN df_manufacturers_clean m ON p.part_id = m.part_id
            WHERE m.origin_country = '{selected_country}'
            """

            eredmeny = duckdb.query(query).to_df()
            display(eredmeny)

# Figyeljük a változást
dropdown.observe(on_change)

# Megjelenítjük a menüt és a táblázat helyét
display(dropdown)
display(output)

# Az első indításhoz "megpöccintjük" a függvényt
on_change({'type': 'change', 'name': 'value', 'new': countries[0]})
```
